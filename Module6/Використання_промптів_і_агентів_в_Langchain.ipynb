### Завдання 2: Створення параметризованого промпта для генерації тексту
Тепер ми хочемо оновити попередній фукнціонал так, аби в промпт ми могли передавати тему як параметр. Для цього скористайтесь `PromptTemplate` з `langchain` і реалізуйте параметризований промпт та виклик моделі з ним.

Запустіть оновлений функціонал (промпт + модел) для пояснень про теми
- "Баєсівські методи в машинному навчанні"

Виведіть результати відпрацювання моделі на екран.

````python
from langchain_core.prompts import PromptTemplate

prompt = PromptTemplate(
    input_variables=["topic"],
    template="Explain what {topic} is shortly and in simple terms using definition, main advantages and current investigations in this field.",
)

print(llm.invoke(prompt.format(topic="Bayesian methods in machine learning")))
print(llm.invoke(prompt.format(topic="Transformers in machine learning")))
print(llm.invoke(prompt.format(topic="Explainable A")))
````

````python
from langchain_classic.agents import load_tools, Tool, AgentExecutor, AgentType, create_react_agent, initialize_agent
from langchain_core.prompts import ChatPromptTemplate
from langchain_community.tools import DuckDuckGoSearchRun
````

````python
web_search = DuckDuckGoSearchRun()
search_tool = Tool(
    name="Web Search",
    func=web_search.run,
    description="Useful for searching the internet for recent information, articles, and scientific papers on a given topic."
)
````

````python
tools = [search_tool]
tools
````

````python
from langchain_core.prompts import PromptTemplate

react_prompt_template = """You are a helpful research assistant. Your task is to find information about recent scientific publications on a given topic.

Use the following format:

Question: the input question you must answer
Thought: you should always think about what to do
Action: the action to take, should be one of [{tool_names}]
Action Input: the input to the action
Observation: the result of the action
... (this Thought/Action/Action Input/Observation can repeat N times)
Thought: I now know the final answer
Final Answer: the final answer to the original input question, structured as a clean list of 5 recent scientific publications (including Title, Authors, and a short summary).

Begin!

Tools available:
{tools}

Using tools names: {tool_names}

Question: {input}
Thought:{agent_scratchpad}"""

prompt = PromptTemplate.from_template(react_prompt_template)
````

````python
agent = create_react_agent(llm, tools, prompt)
agent_executor = AgentExecutor(
    agent=agent, tools=tools, verbose=True, handle_parsing_errors=True
)

response = agent_executor.invoke({
    "input": (
        "Find 5 latest scientific publications or papers about cats, including title, authors, and a short summary for each."
    )
})

print(response["output"])
````

````python
from langchain_experimental.utilities import PythonREPL
````

````python
python_repl = PythonREPL()
python_tool = Tool(
    name="python_interpreter",
    func=python_repl.run,
    description="Use for any data analysis, complex calculations, prediction or logic and process statistics. Input must be valid Python code that uses print() to output the result."
)
````

````python
tools = [search_tool, python_tool]
````

````python
prompt_template = """You are a helpful assistant for solving business problem. Your task is to work with analytical data and help solve business analytics problems.

Use the following format:

Question: the input question you must answer
Thought: you should always think about what to do
Action: the action to take, should be one of [{tool_names}]
Action Input: the input to the action
Observation: the result of the action
... (this Thought/Action/Action Input/Observation can repeat N times)
Thought: I now know the final answer
Final Answer: the final answer to the original input question, providing a clear sales forecast for the next year based on the data, inflation, and weather conditions.

Begin!

Tools available:
{tools}

Using tools names: {tool_names}

Question: {input}
Thought:{agent_scratchpad}"""

prompt = PromptTemplate.from_template(prompt_template)
````

````python
agent = create_react_agent(llm, tools, prompt)
agent_executor = AgentExecutor(
    agent=agent, tools=tools, verbose=True, handle_parsing_errors=True
)

response = agent_executor.invoke({
    "input": (
        "Тобі потрібно видати результат бізнес аналізу моєї задачі. Якщо ти не знаєш відповіді напиши що ти не міг вирішити цю задачу та поверни оригінальний текст задачі із запиту користувача.  Задача - Ми експортуємо апельсини з Бразилії. В 2022 експортували 200т, в 2023 - 190т, в 2024 - 210т, в 2025 - 220т. Зроби оцінку скільки ми зможемо експортувати апельсинів в 2026 враховуючи погодні умови в Бразилії і попит на апельсини в світі виходячи з економічної ситуації. "
    )
})

print(response["output"])
````
